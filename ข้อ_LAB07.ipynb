```python
import numpy as np # linear algebra
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
enroll=pd.read_excel('/content/drive/MyDrive/LabDataScince/enrollement.xlsx')
enroll.head(20)
```

```python
course=pd.read_csv('/content/drive/MyDrive/LabDataScince/course.csv')
course.head(15)
```

```python
course_enroll = pd.merge(course, enroll)
course_enroll
```

```python
enrolls_matrix = enroll.pivot_table(index=['COURSECODE'],columns=['STUDENTCODE'],values='GRADE').reset_index(drop=True)
enrolls_matrix.fillna( 0, inplace = True )
enrolls_matrix
```

```python
course_similarity=cosine_similarity(enrolls_matrix)
np.fill_diagonal(course_similarity, 0 )
enrolls_matrix = pd.DataFrame( course_similarity )
enrolls_matrix.head(15)
```

```python
try:
    course_inp=input('Enter the reference course title based on which recommendations are to be made: ')
    inp=course[course['COURSENAME']==course_inp].index.tolist()
    inp=inp[0]

    course['similarity'] = enrolls_matrix.iloc[inp]
    course.head(5)

except:
    print("Sorry, the course is not in the database!")

print("Recommended course ",course_inp ,"\n", course.sort_values( ["similarity"], ascending = False )[0:10])
```

```python
def get_recommendations(std_id, top_n=10):
    try:
        std_enroll = enroll[enroll['STUDENTCODE'] == std_id]
        if std_enroll.empty:
          return "User not found in the dataset."
        std_course_ids = std_enroll['COURSECODE'].unique()
        # Calculate average similarity for movies the user has rat๐ed
        similarities = []
        for course_id in std_course_ids:
            course_index = course[course['COURSECODE'] == course_id].index[0]  # Get the index
            similarities.append(enrolls_matrix.iloc[course_index])

        average_similarity = pd.concat(similarities, axis=1).mean(axis=1)

        # Sort movies based on average similarity and recommend top_n movies
        course['similarity'] = average_similarity
        recommendations = course.sort_values(by=['similarity'], ascending=False)
        return recommendations.head(top_n)

    except Exception as e:
        return f"An error occurred: {e}"

# Get recommendations for a specific user
std_id_input = input("Enter the student ID: ")
recommendations = get_recommendations(std_id_input)
recommendations
```
